Это чтобы текст для зашифровки получить (нужен интернет)

```python
from bs4 import BeautifulSoup
import requests
import re

def removeHTML(html_text: str):
    clean_text = re.sub(r'<[^>]+>', '', html_text)
    return clean_text.replace("\r\n", "")

def getText4Encrypt() -> str:
    result = ""
    response = requests.get('https://www.trees-and-lambdas.info/matushansky/stirlitz.html')
    soup = BeautifulSoup(response.content, 'html.parser')
    c = 0
    for text in soup.find_all("p"):
        if (c == 0): c += 1
        else: result += removeHTML(text.text) + '\n' * 2
    return result
```
